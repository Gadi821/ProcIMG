Solarizacion

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def solarizar(imagen, umbral):
    img_solarizada = np.where(imagen<umbral,imagen,255-imagen)
    return img_solarizada

imagen = cv2.imread('imgGIF3.jpg')
img_gris = cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
umbral = 127
img_nueva = solarizar(img_gris,umbral)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img solarizada')
plt.imshow(img_nueva,cmap='gray')
plt.axis('off')

plt.show()
```

Posterizacion

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def posterizar(imagen,nivel):
    factor = 255/nivel
    img_post = np.round((imagen/factor)*factor)
    return img_post.astype(np.uint8)

imagen = cv2.imread('imgGIF3.jpg')
img_gris = cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
nivel = 6 #nivel de intencidad
img_nueva = posterizar(img_gris,nivel)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img posterizada')
plt.imshow(img_nueva,cmap='gray')
plt.axis('off')

plt.show()
```

Falso Color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def falso_color(imagen_gray):
    escala_imagen = cv2.normalize(imagen_gray,None,0,255,cv2.NORM_MINMAX)
    apli_color = cv2.applyColorMap(escala_imagen,cv2.COLORMAP_INFERNO)
    return apli_color


imagen = cv2.imread('imgGIF3.jpg')
img_gris = cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)

img_nueva = falso_color(img_gris)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img con falso color')
plt.imshow(cv2.cvtColor(img_nueva,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

Umbralizaciones por defecto tuzero en opencv

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imgGIF3.jpg',0)
_,umbral_trunc=cv2.threshold(imagen,127,255,cv2.THRESH_TRUNC)
_,umbral_tozero=cv2.threshold(imagen,127,255,cv2.THRESH_TOZERO)
_,umbral_tozero_INV=cv2.threshold(imagen,127,255,cv2.THRESH_TOZERO_INV)

plt.figure(figsize=(12,5))
plt.subplot(1,4,1)
plt.imshow(imagen,cmap='gray')
plt.title('imagen original')

plt.subplot(1,4,2)
plt.imshow(umbral_trunc,cmap='gray')
plt.title('imagen original')
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(umbral_tozero,cmap='gray')
plt.title('imagen original')
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(umbral_tozero_INV,cmap='gray')
plt.title('imagen original')
plt.axis('off')

plt.show()
```

Negativo de una imagen

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imgGIF3.jpg')
img_negativo=255-imagen

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.title('imagen original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_negativo,cv2.COLOR_BGR2RGB))
plt.title('imagen negativo')
plt.axis('off')

plt.show()
```
