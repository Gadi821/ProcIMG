Convolucion

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('imgGIF3.jpg',0)
mascara = np.array([[0,-1,0],[1,5,-1],[0,1,0]]) #kernel
img_fin = cv2.filter2D(img,-1,mascara)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img concolucion')
plt.imshow(img_fin, cmap='gray')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('imgGIF3.jpg')
kernel5x5 = np.ones((5,5),np.float32)/25 #suavizado
suavizado = cv2.filter2D(img,-1,kernel5x5) #img suavizado

kernel7x7 =  np.ones((7,7),np.float32)/49 #suavizado fuerte
suavizadoF2 = cv2.filter2D(img,-1,kernel7x7)

plt.figure(figsize=(10,10))
plt.subplot(1,3,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,3,2)
plt.title('img kernel5x5')
plt.imshow(cv2.cvtColor(suavizado,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,3,3)
plt.title('img kernel7x7')
plt.imshow(cv2.cvtColor(suavizadoF2,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_horiz(img,tamKernel):
    kernel = np.ones((1,tamKernel),np.float32)/tamKernel
    imgfin = cv2.filter2D(img,-1,kernel)
    return imgfin

img = cv2.imread('imgDistor2.jpg')
imgRGB = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

imgResR = convolucion_media_horiz(imgRGB[:,:,0],25)
imgResG = convolucion_media_horiz(imgRGB[:,:,1],25)
imgResB = convolucion_media_horiz(imgRGB[:,:,2],25)

imgColor = np.stack([imgResR,imgResG,imgResB],axis=2)

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img kernel1x5')
plt.imshow(imgColor)
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_horiz(img,tamKernel):
    kernel = np.ones((tamKernel,1),np.float32)/tamKernel
    imgfin = cv2.filter2D(img,-1,kernel)
    return imgfin

img = cv2.imread('imgDistor2.jpg')
imgRGB = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

imgResR = convolucion_media_horiz(imgRGB[:,:,0],25)
imgResG = convolucion_media_horiz(imgRGB[:,:,1],25)
imgResB = convolucion_media_horiz(imgRGB[:,:,2],25)

imgColor = np.stack([imgResR,imgResG,imgResB],axis=2)

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img kernel1x5')
plt.imshow(imgColor)
plt.axis('off')

plt.show()
```

Ruido

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img_ruido = np.random.randint(0,256,size=(256,256),dtype=np.uint8)
img_blur = cv2.GaussianBlur(img_ruido,(5,5),0)

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.title('img ruido')
plt.imshow(img_ruido,cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img blur')
plt.imshow(img_blur, cmap='gray')
plt.axis('off')

plt.show()
```
