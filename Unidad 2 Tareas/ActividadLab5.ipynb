Filtros Bayer

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imag1JPEG.jpg', 0)
imagen_B1 = cv2.cvtColor(imagen, cv2.COLOR_BayerBG2BGR)
imagen_B2 = cv2.cvtColor(imagen, cv2.COLOR_BayerGB2BGR)
imagen_B3 = cv2.cvtColor(imagen, cv2.COLOR_BayerRG2BGR)
imagen_B4 = cv2.cvtColor(imagen, cv2.COLOR_BayerGR2BGR)

plt.figure(figsize=(10, 5))
plt.subplot(1, 5, 1)
plt.title('imagen original')
plt.imshow(imagen)
plt.axis('off')

plt.subplot(1, 5, 2)
plt.title('(BG2BGR)')
plt.imshow(cv2.cvtColor(imagen_B1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 5, 3)
plt.title('(GB2BGR)')
plt.imshow(cv2.cvtColor(imagen_B2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 5, 4)
plt.title('(RG2BGR)')
plt.imshow(cv2.cvtColor(imagen_B3, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 5, 5)
plt.title('(GR2BGR)')
plt.imshow(cv2.cvtColor(imagen_B4, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

TRansformaciones YUV2RGB RGB2YUB

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imag1JPEG.jpg', 0)
imagen_Y1 = cv2.cvtColor(imagen, cv2.COLOR_BGR2YUV)
imagen_Y2 = cv2.cvtColor(imagen, cv2.COLOR_YUV2BGR)

plt.figure(figsize=(10, 5))
plt.subplot(1, 3, 1)
plt.title('imagen original')
plt.imshow(imagen)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title('(BG2BGR)')
plt.imshow(cv2.cvtColor(imagen_Y1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title('(GB2BGR)')
plt.imshow(cv2.cvtColor(imagen_Y2, cv2.COLOR_BGR2RGB))
plt.axis('off')
```
