Cargar 3 fotos

```python
import cv2
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=[10,5]
img1=cv2.imread('Foto1.jpg',cv2.IMREAD_COLOR)
img2=cv2.imread('Foto2.jpg',cv2.IMREAD_COLOR)
img3=cv2.imread('Foto3.jpg',cv2.IMREAD_COLOR)

plt.subplot(1,3,1)
plt.title('Foto 1')
plt.imshow(cv2.cvtColor(img1,cv2.COLOR_BGR2RGB))

plt.subplot(1,3,2)
plt.title('Foto 2')
plt.imshow(cv2.cvtColor(img2,cv2.COLOR_BGR2RGB))

plt.subplot(1,3,3)
plt.title('Foto 3')
plt.imshow(cv2.cvtColor(img3,cv2.COLOR_BGR2RGB))

plt.show()
```

Funcion mostrar imagen

```python
import cv2
import matplotlib.pyplot as plt
def MostrarImagen(x):
    imagen=cv2.imread(x)
    imagrgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

    plt.title("importada imagen")
    plt.imshow(imagrgb)
    plt.axis('off')
    plt.show()  

MostrarImagen('foto3.jpg')
MostrarImagen('foto2.jpg')
MostrarImagen('Foto1.jpg')
```

Fuciones de modos de subplot

Escala de grises

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def EscalaGrises(x):
    plt.rcParams['figure.figsize']=[10,5]
    img1=cv2.imread(x,cv2.IMREAD_COLOR)

    plt.subplot(1,4,1)
    plt.title('Imagen Original')
    plt.imshow(cv2.cvtColor(img1,cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.subplot(1,4,2)
    plt.title('canal rojo')
    plt.imshow(img1[:,:,0],cmap='gray')
    plt.axis('off')

    plt.subplot(1,4,3)
    plt.title('canal verde')
    plt.imshow(img1[:,:,1],cmap='gray')
    plt.axis('off')

    plt.subplot(1,4,4)
    plt.title('canal azul')
    plt.imshow(img1[:,:,2],cmap='gray')
    plt.axis('off')

    plt.show()
```

Canales / Colores originales

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def FormatoColor(y):
    plt.rcParams['figure.figsize']=[10,5]
    img=cv2.imread(y,cv2.IMREAD_COLOR)
    img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

    r=img[:,:,2]
    g=img[:,:,1]
    b=img[:,:,0]

    r_imagen=np.zeros_like(img_rgb)
    r_imagen[:,:,0]=r

    g_imagen=np.zeros_like(img_rgb)
    g_imagen[:,:,1]=g

    b_imagen=np.zeros_like(img_rgb)
    b_imagen[:,:,2]=b

    plt.subplot(1,4,1)
    plt.title('Imagen Original')
    plt.imshow(img_rgb)
    plt.axis('off')

    plt.subplot(1,4,2)
    plt.title('canal rojo')
    plt.imshow(r_imagen)
    plt.axis('off')

    plt.subplot(1,4,3)
    plt.title('canal verde')
    plt.imshow(g_imagen)
    plt.axis('off')

    plt.subplot(1,4,4)
    plt.title('canal azul')
    plt.imshow(b_imagen)
    plt.axis('off')

    plt.show()

FormatoColor('Foto1.jpg')
```

Canales / Colores Falsos

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def ColoresFalsos(z):
    plt.rcParams['figure.figsize']=[10,5]
    img=cv2.imread(z,cv2.IMREAD_COLOR)
    b,g,r=cv2.split(img)

    plt.subplot(1,4,1)
    plt.title('Imagen Original')
    plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
    plt.axis('off')

    plt.subplot(1,4,2)
    plt.title('canal rojo')
    plt.imshow(r,cmap='Reds')
    plt.axis('off')

    plt.subplot(1,4,3)
    plt.title('canal verde')
    plt.imshow(g,cmap='Greens')
    plt.axis('off')

    plt.subplot(1,4,4)
    plt.title('canal azul')
    plt.imshow(b,cmap='Blues')
    plt.axis('off')

    plt.show()

ColoresFalsos('Foto1.jpg')
```

Segmentos de las fotos

```python
def SegmentarImg(x, num1, num2,titulo):
    img1=cv2.imread(x,cv2.IMREAD_COLOR)

    plt.imshow(img1[:num1,:num2,:])
    plt.title(titulo)
    plt.show()


SegmentarImg('Foto1.jpg',50,50,'Foto 1')
SegmentarImg('foto2.jpg',50,50,'Foto 2')
SegmentarImg('foto3.jpg',50,50,'Foto 3')
```
