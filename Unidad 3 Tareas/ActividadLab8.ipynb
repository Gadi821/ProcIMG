Matrices de Convolucion

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def Enfoque(img):
    mascara = np.array([[0,-1,0],[-1,5,-1],[0,1,0]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin

def Desenfoque(img):
    mascara = np.array([[1,1,1],[1,1,1],[1,1,1]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin

def RealceBordes(img):
    mascara = np.array([[0,0,0],[-1,1,0],[0,0,0]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin

def Repujado(img):
    mascara = np.array([[-2,-1,0],[-1,1,1],[0,1,2]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin

def DeteccionBordes(img):
    mascara = np.array([[0,1,0],[1,-4,1],[0,1,0]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin

def FiltroSobel(img):
    mascara = np.array([[-1,0,1],[-2,0,2],[-1,0,1]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin

def FiltroSharper(img):
    mascara = np.array([[1,-2,1],[-2,5,-2],[1,-2,1]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin

def FiltroNorte(img):
    mascara = np.array([[1,1,1],[1,-2,1],[-1,-1,-1]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin

def FiltroEste(img):
    mascara = np.array([[-1,1,1],[-1,-2,1],[-1,1,1]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin

def FiltroGauss(img):
    mascara = np.array([[1,2,3,1,1],[2,7,11,7,2],[3,11,17,11,3],[2,7,11,7,2],[1,2,3,2,1]]) #kernel
    img_fin = cv2.filter2D(img,-1,mascara)
    return img_fin
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('imgGIF3.jpg')

img_enfoque = Enfoque(img)
img_desenfoque = Desenfoque(img)
img_realce_bordes = RealceBordes(img)
img_repujado = Repujado(img)
img_deteccion_bordes = DeteccionBordes(img)
img_sobel = FiltroSobel(img)
img_sharper = FiltroSharper(img)
img_norte = FiltroNorte(img)
img_este = FiltroEste(img)
img_gauss = FiltroGauss(img)

plt.figure(figsize=(15, 12))

plt.subplot(3, 4, 1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(3, 4, 2)
plt.imshow(cv2.cvtColor(img_enfoque,cv2.COLOR_BGR2RGB))
plt.title('Enfoque')
plt.axis('off')

plt.subplot(3, 4, 3)
plt.imshow(cv2.cvtColor(img_desenfoque,cv2.COLOR_BGR2RGB))
plt.title('Desenfoque')
plt.axis('off')

plt.subplot(3, 4, 4)
plt.imshow(cv2.cvtColor(img_realce_bordes,cv2.COLOR_BGR2RGB))
plt.title('Realce Bordes')
plt.axis('off')

plt.subplot(3, 4, 5)
plt.imshow(cv2.cvtColor(img_repujado,cv2.COLOR_BGR2RGB))
plt.title('Repujado')
plt.axis('off')

plt.subplot(3, 4, 6)
plt.imshow(cv2.cvtColor(img_deteccion_bordes,cv2.COLOR_BGR2RGB))
plt.title('Detección Bordes')
plt.axis('off')

plt.subplot(3, 4, 7)
plt.imshow(cv2.cvtColor(img_sobel,cv2.COLOR_BGR2RGB))
plt.title('Filtro Sobel')
plt.axis('off')

plt.subplot(3, 4, 8)
plt.imshow(cv2.cvtColor(img_sharper,cv2.COLOR_BGR2RGB))
plt.title('Filtro Sharper')
plt.axis('off')

plt.subplot(3, 4, 9)
plt.imshow(cv2.cvtColor(img_norte,cv2.COLOR_BGR2RGB))
plt.title('Filtro Norte')
plt.axis('off')

plt.subplot(3, 4, 10)
plt.imshow(cv2.cvtColor(img_este,cv2.COLOR_BGR2RGB))
plt.title('Filtro Este')
plt.axis('off')

plt.subplot(3, 4, 11)
plt.imshow(cv2.cvtColor(img_gauss,cv2.COLOR_BGR2RGB))
plt.title('Filtro Gauss')
plt.axis('off')

plt.tight_layout()
plt.show()
```
