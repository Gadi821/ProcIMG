```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_horiz(img,tamKernel):
    kernel = np.ones((1,tamKernel),np.float32)/tamKernel
    imgfin = cv2.filter2D(img,-1,kernel)
    return imgfin

def convolucion_media_vert(img,tamKernel):
    kernel = np.ones((tamKernel,1),np.float32)/tamKernel
    imgfin = cv2.filter2D(img,-1,kernel)
    return imgfin

def convolucion_media_diag_izdader(img, tamKernel):
    kernel = np.zeros((tamKernel, tamKernel), np.float32)
    np.fill_diagonal(kernel, 1) #diagonal izq a der
    kernel /= tamKernel  
    imgfin = cv2.filter2D(img, -1, kernel)
    return imgfin

def convolucion_media_diag_deraizq(img, tamKernel):
    kernel = np.zeros((tamKernel, tamKernel), np.float32)
    kernel = np.fliplr(np.eye(tamKernel))  # diagonal der a izd
    kernel /= tamKernel
    imgfin = cv2.filter2D(img, -1, kernel)
    return imgfin


img = cv2.imread('imgLab91.jpg')
imgRGB = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

img1 = convolucion_media_horiz(img,25)
img2 = convolucion_media_vert(img1,25)
img3 = convolucion_media_diag_izdader(img,25)
img4 = convolucion_media_diag_deraizq(img,25)

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img vert-horizontal')
plt.imshow(cv2.cvtColor(img2,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.title('img diagonal izq a der')
plt.imshow(cv2.cvtColor(img3,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img diagonal der a izq')
plt.imshow(cv2.cvtColor(img4,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```
