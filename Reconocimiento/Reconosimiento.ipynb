Reconocimiento Parte1 Toma de imagenes

```python
import cv2
import os

import cv2.data

nombre = "nom_persona"
ruta=f'dataset/{nombre}'
os.makedirs(ruta, exist_ok=True)
captura = cv2.VideoCapture(0)
face_cascada = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

cont = 0
while True:
    retrato, frames = captura.read()
    if not retrato:
        break

    gray = cv2.cvtColor(frames,cv2.COLOR_BGR2GRAY)
    faces = face_cascada.detectMultiScale(gray,1.3,5)
    for(x,y,w,h) in faces:
        rostro = gray[y:y-h,x:x-w]
        cv2.imwrite(f'{ruta}/{cont}.jpg',rostro)
        cont+=1
        cv2.rectangle(frames,(x,y),(x+w,y+h),(255,0,0),2)

    cv2.imshow('captura',frames)
    if (cv2.waitKey(1)==27 or cont >= 20):
        break

captura.release()
cv2.destroyAllWindows()
```
