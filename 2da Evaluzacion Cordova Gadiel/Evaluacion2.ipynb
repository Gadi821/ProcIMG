```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('ruido.jpg')
img_gray = cv2.Laplacian(img, cv2.CV_64F)
suabizado = cv2.GaussianBlur(img_gray,(3,3),0)

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img Empañada')
plt.imshow(suabizado)
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('cristo.jpg')

kernel7x7 =  np.ones((7,7),np.float32)/49 #suavizado fuerte
suavizadoF2 = cv2.filter2D(img,-1,kernel7x7)

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img Empañada')
plt.imshow(cv2.cvtColor(suavizadoF2,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```
