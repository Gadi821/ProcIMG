```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

def blurN(img,mask,blur):
    fondo = cv2.GaussianBlur(img,(51,51),0)
    fin = np.where(mask[:,:,None]>0,img,fondo)
    return fin

imagen = cv2.imread('rosa.png')
imgGray = cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)

_,mascara = cv2.threshold(imgGray,100,125,cv2.THRESH_BINARY)
mascara = cv2.bitwise_not(mascara)
mascara = cv2.GaussianBlur(mascara,(21,21),0)
blur = 25
res = blurN(imagen,mascara,blur)

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img Empañada')
plt.imshow(res)
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

imagen = cv2.imread('elementos.jpg')
imgGray = cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
_,binario=cv2.threshold(imgGray,128,255,cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
borde,_ = cv2.findContours(binario,cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
cont = len(borde)

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.title(f'cantidad de objetos: {cont}')
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,2,2)
plt.title(f'cantidad de objetos: {cont}')
plt.imshow(cv2.cvtColor(imgGray,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```
