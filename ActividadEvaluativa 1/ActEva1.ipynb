```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

colores=[(0,'purple'),(0.2,'darkblue'),(0.8,'green'),(1,'yellow')]
cmap = LinearSegmentedColormap.from_list('degradado',colores)
imagen_degra = cmap(np.linspace(0,1,650).reshape(-1,1))

plt.imshow(imagen_degra,aspect='auto')
plt.show()
```

Ejercicio 3

```python
import matplotlib.pyplot as plt
import numpy as np

tamaño_img = 100
R = np.ones((tamaño_img,tamaño_img))
G = np.ones((tamaño_img,tamaño_img))
B = np.ones((tamaño_img,tamaño_img))

blanco = np.dstack((R * 255,G * 255,B * 255))  

negro = np.dstack((R * 0,G * 0,B * 0))     


superior = np.hstack((negro, blanco))
inferior = np.hstack((blanco, negro))
#horizontal
imagenh1 = np.vstack((superior, inferior))
imagenh2 = np.hstack((imagenh1,imagenh1))
imagenh3 = np.hstack((imagenh2,imagenh2))
#vertical
imagenv3 = np.vstack((imagenh3,imagenh3))
imagenx = np.vstack((imagenv3,imagenv3))
plt.imshow(imagenx.astype(int))
plt.show()
```

Ejercicio 5

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize']=[10,5]
imagen=cv2.imread('imagen1.jpg',0)


degrad_lineal= np.linspace(1,-0.1,imagen.shape[0])
mask1 = np.tile(degrad_lineal, (imagen.shape[1], 1)).T#crear mascara y dar vuelta al delineado
normalizar = cv2.normalize(imagen, None, 0, 1, cv2.NORM_MINMAX) #juntar y normalizar
img_degradada = normalizar * mask1


plt.subplot(1,2,1)
plt.imshow(imagen,cmap='gray')
plt.title('imagen original')

plt.subplot(1,2,2)
plt.imshow(img_degradada,cmap='gray')
plt.title('imagen ajustada')
plt.axis('off')

plt.show()
```
