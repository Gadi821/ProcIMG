Ecualizacion de un Histograma - Normalizar a un punto optimo

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imgGIF3.jpg',0)

histograma,calc=np.histogram(imagen.flatten(),256,[0,256])
distriAcu = histograma.cumsum()
distriAcu_normalizada = distriAcu*histograma.max()/distriAcu.max()
ecualizado = cv2.equalizeHist(imagen)
histogrma_equ,_=np.histogram(ecualizado.flatten(),256,[0,256])

plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.imshow(imagen,cmap='gray')
plt.title('imagen original')
plt.axis('off')

plt.subplot(2,2,2)
plt.plot(distriAcu_normalizada,color='b')
plt.hist(imagen.flatten(),256,[0,256],color='r')
plt.xlim([0,256])
plt.legend(('distribucion','histograma'),loc='upper left')
plt.title('histograma original')


plt.subplot(2,2,3)
plt.imshow(ecualizado,cmap='gray')
plt.title('imagen ecualizada')
plt.axis('off')

plt.subplot(2,2,4)
plt.plot(histogrma_equ,color='b')
plt.xlim([0,256])
plt.tight_layout()

plt.show()
```

ecualizacion a color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imgGIF3.jpg')
imagen_gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
ecualizado = cv2.equalizeHist(imagen_gris)
ecu_color = cv2.merge([cv2.equalizeHist(imagen[:,:,0]),cv2.equalizeHist(imagen[:,:,1]),cv2.equalizeHist(imagen[:,:,2])])

plt.imshow(ecualizado,cmap='gray')
plt.show()

plt.imshow(cv2.cvtColor(ecu_color,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imgGIF3.jpg',0)
clip = 2.0
tiles = 8

clahe = cv2.createCLAHE(clipLimit=clip,tileGridSize=(tiles,tiles))
ecualizado = clahe.apply(imagen)

plt.imshow(imagen,cmap='gray')
plt.show()

plt.imshow(ecualizado,cmap='gray')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imgGIF3.jpg',0)

trasnformada = np.fft.fft2(imagen) #aplicar trasformada e inversa de forier
desplazar = np.fft.fftshift(trasnformada) #separo el espectro
magnitudEspectro = np.log(1+np.abs(desplazar)) #calculo de espectro separado

plt.imshow(imagen,cmap='gray')
plt.show()

plt.imshow(magnitudEspectro,cmap='gray')
plt.show()
```
