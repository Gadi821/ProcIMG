```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('imgGIF3.jpg',cv2.IMREAD_GRAYSCALE)
imgRGB = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

#elemento estructurante
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(5,5))
#aplicar erocion
imgRest = cv2.erode(img,kernel)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title('img original')
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(1,2,2)
plt.title('img erosionada')
plt.imshow(imgRest,cmap='gray')
plt.axis('off')

plt.show()
```

utilizar skimagen con erocion dilatacion apertura y cerradura

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import morphology
 
imagen = cv2.imread('imgGIF3.jpg', cv2.IMREAD_GRAYSCALE)
_, imagen_bin = cv2.threshold(imagen, 127, 255, cv2.THRESH_BINARY)

kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(9,9))

erosion_img = morphology.erosion(imagen_bin, kernel)
dilatacion_img = morphology.dilation(imagen_bin, kernel)
apertura_img = morphology.opening(imagen_bin, kernel)
cerradura_img = morphology.closing(imagen_bin, kernel)
 
# Mostrar los resultados
plt.figure(figsize=(10, 5))
 
plt.subplot(1, 5, 1)
plt.imshow(imagen, cmap='gray')
plt.title('Original')
plt.axis('off')
 
plt.subplot(1, 5, 2)
plt.imshow(erosion_img, cmap='gray')
plt.title('Erosión')
plt.axis('off')
 
plt.subplot(1, 5, 3)
plt.imshow(dilatacion_img, cmap='gray')
plt.title('Dilatación')
plt.axis('off')
 
plt.subplot(1, 5, 4)
plt.imshow(apertura_img, cmap='gray')
plt.title('Apertura')
plt.axis('off')
 
plt.subplot(1, 5, 5)
plt.imshow(cerradura_img, cmap='gray')
plt.title('Cerradura')
plt.axis('off')
 
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.data import checkerboard

#Imagen con pixeles
imagen = np.zeros((300, 300), dtype=np.uint8)
#
imagen = cv2.circle(imagen, (200, 100), 40, 255, -1)
imagen = cv2.rectangle(imagen, (50, 50), (100, 100), 255, -1)
imagen = cv2.line(imagen, (250, 50), (280, 200), 255, 10)

check = checkerboard()>8.5
check = check.astype(np.uint8)*255

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(imagen, cmap='gray')
plt.title('img')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(check, cmap='gray')
plt.title('checkerboard')
plt.axis('off')


plt.show()

kernel_ellipse = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9, 9))
kernel_cross = cv2.getStructuringElement(cv2.MORPH_CROSS, (9, 9))
kernel_rect = cv2.getStructuringElement(cv2.MORPH_RECT, (9, 9))

#Aplicando
erode_ellipse = cv2.erode(imagen, kernel_ellipse,iterations=1)
dilate_ellipse = cv2.dilate(imagen, kernel_ellipse, iterations=1)

erode_croos = cv2.erode(imagen, kernel_cross,iterations=1)
dilate_croos = cv2.dilate(imagen, kernel_cross, iterations=1)

erode_rect = cv2.erode(imagen, kernel_rect,iterations=1)
dilate_rect = cv2.dilate(imagen, kernel_rect, iterations=1)

erode_check = cv2.erode(check, kernel_ellipse,iterations=1)
dilate_check = cv2.dilate(check, kernel_ellipse, iterations=1)

erode_check = cv2.erode(check, kernel_rect,iterations=1)
dilate_check = cv2.dilate(check, kernel_rect, iterations=1)

erode_check = cv2.erode(check, kernel_cross,iterations=1)
dilate_check = cv2.dilate(check, kernel_cross, iterations=1)


#Creamos imagen apertura y cerradura
np.random.seed(42)
imagen_ruido = imagen.copy()
ruido = np.random.randint(0, 2, imagen.shape, dtype=np.uint8)*255
#Juntar Excluyente
imagen_ruido = cv2.bitwise_xor(imagen_ruido, ruido)
apertura = cv2.morphologyEx(imagen_ruido, cv2.MORPH_OPEN, kernel_rect)
cerradura = cv2.morphologyEx(imagen_ruido, cv2.MORPH_CLOSE, kernel_rect)

plt.figure(figsize=(10, 5))

plt.subplot(3, 4, 1)
plt.imshow(erode_ellipse, cmap='gray')
plt.title('Erode ELLIPSE')
plt.axis('off')

plt.subplot(3, 4, 2)
plt.imshow(dilate_ellipse, cmap='gray')
plt.title('Dilate ELLIPSE')
plt.axis('off')

plt.subplot(3, 4, 3)
plt.imshow(erode_croos, cmap='gray')
plt.title('Erode Cross')
plt.axis('off')

plt.subplot(3, 4, 4)
plt.imshow(dilate_croos, cmap='gray')
plt.title('Erode Cross')
plt.axis('off')

plt.subplot(3, 4, 5)
plt.imshow(erode_rect, cmap='gray')
plt.title('Erode Rect')
plt.axis('off')

plt.subplot(3, 4, 6)
plt.imshow(dilate_rect, cmap='gray')
plt.title('Dilate Rect')
plt.axis('off')

plt.subplot(3, 4, 7)
plt.imshow(erode_check, cmap='gray')
plt.title('Erode Check')
plt.axis('off')


plt.subplot(3, 4, 8)
plt.imshow(dilate_check, cmap='gray')
plt.title('Dilate Check')
plt.axis('off')

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(apertura, cmap='gray')
plt.title('apertura')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cerradura, cmap='gray')
plt.title('cerradura')
plt.axis('off')


plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.morphology import reconstruction

img = cv2.imread('imgGIF3.jpg',cv2.IMREAD_GRAYSCALE)
_,imgBin = cv2.threshold(img,127,255,cv2.THRESH_BINARY)

kernel_ellipse = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9, 9))
kernel_cross = cv2.getStructuringElement(cv2.MORPH_CROSS, (9, 9))
kernel_rect = cv2.getStructuringElement(cv2.MORPH_RECT, (9, 9))

hitormiss1 = cv2.morphologyEx(imgBin,cv2.MORPH_HITMISS,kernel_ellipse) 

#algoritmo de detecion de cruces
kernel_L = np.array([
    [0, 1, 0],
    [0, 1, 0],
    [0, 1, 1]
], dtype=np.uint8)

hitormiss_L = cv2.morphologyEx(imgBin, cv2.MORPH_HITMISS, kernel_L)

plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('img')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(hitormiss1, cmap='gray')
plt.title('hitormiss')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(hitormiss_L, cmap='gray')
plt.title('hitormiss en L')
plt.axis('off')

plt.show()

#transformaciones geodesicas
img2 = np.copy(imgBin)
mask = imgBin

ero_geodesica = reconstruction(img2,mask,method='erosion')
dila_geodesica = reconstruction(img2,mask,method='dilation')

plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('img')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(ero_geodesica, cmap='gray')
plt.title('ero_geodesica')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(dila_geodesica, cmap='gray')
plt.title('dila_geodesica')
plt.axis('off')

plt.show()
```
